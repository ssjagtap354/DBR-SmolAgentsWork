### we are creating this to mimic how task values are passed in between 
- we will fetch all categories from bronze 
- then we will process each categor by creating seperate partition of it and storing in new schema

```python
df = spark.sql("select * from  ecommerce.bronze.brz_category ")
# another way to read managed table 
# df = spark.read.format("delta").table("ecommerce.bronze.brz_category")
df.printSchema()
```

```python
from pyspark.sql.functions import col

df_cat = df.select(col("category_name")).distinct().collect()
cat_list = [each [0] for each in df_cat]


print(f"cat to process are {cat_list}")
```

#### Set the value to taskvalues

```python
dbutils.jobs.taskValues.set(key='cat_names', value=cat_list)
```
