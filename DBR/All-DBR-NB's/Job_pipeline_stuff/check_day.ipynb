#### Create Widget 

```python
dbutils.widgets.text("input_date", "")
```

#### Store value of widget in variable 

```python
inp_date = dbutils.widgets.get("input_date")
print(inp_date)
```

#### Here spark.sql will give dataframe as return type 
- so to extract integer value out of it we use collect method

```python


req_date = spark.sql(f"""
    SELECT date_format(to_date('{inp_date}', 'yyyy-MM-dd'), 'EE') AS day_name
""").collect()[0][0]

print(req_date)
```

##### Set the value to pass it to next task


```python
dbutils.jobs.taskValues.set("run_date",req_date)
```
