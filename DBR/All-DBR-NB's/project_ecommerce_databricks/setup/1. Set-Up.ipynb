```python

%sql
create schema if not exists ecommerce.raw;

-- Volume to drop files from s3
create volume ecommerce.raw.raw_aws_data
```

### Ways to access Data from S3 location in DBR 
- 1. External location
- 2. fs.cp Command
- 3. Copy all files as parquet or Delta. 

### 1. Managed table 

```python
df = spark.read.option("header","True").format("csv").load("s3://sj-dbr-demo-proj/raw_data_landing/brands")
df.show(3)

df.write \
    .format('delta') \
    .saveAsTable("ecommerce.misc.brands_managed")
```

### Using cp command

```python
src_path = "s3://sj-dbr-demo-proj/raw_data_landing/"
dest_path = "/Volumes/ecommerce/raw/raw_aws_data"

dbutils.fs.cp(src_path, dest_path, True)
```
