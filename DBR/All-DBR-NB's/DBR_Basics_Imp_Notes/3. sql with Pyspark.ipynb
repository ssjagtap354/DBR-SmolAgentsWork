### Reading table from catlaog using python interface 

- Now in below 2 methods we are reading table using python approach so we are getting dataframe as return. 
- but what if we just want to query table ? (check cell 2)
    - in this case cell should be %sql and simply write your query as we do in sql
- Suppose we have CSV file which we want to query like SQL 
    - then simply create temp view of that DF and then query it (cell 3) 

- so basically these all are just options. 

```python
df_py_tab = spark.read.table("workspace.default.movies")
# OR 
df_py_tab_2 = spark.sql("select * from workspace.default.movies")
df_py_tab_2.show(2)
```

```python
%sql
select * from workspace.default.movies limit 2 
```

```python
csv_df = spark.read.option("Header",True).csv("/Volumes/workspace/default/raw_data/orders.csv")
csv_df.createOrReplaceTempView('tmp_vw_orders')
```

```python
%sql
select * from tmp_vw_orders
```

### what is _sqldf
- _sqldf exists only in the current notebook
- It holds only the most recent SQL result
- It is temporary (not saved anywhere)
