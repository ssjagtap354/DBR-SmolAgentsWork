### Creating new column (using withColumn)

```python
from pyspark.sql.functions import col
df_with_revenue = df_movies.withColumn('Profit',col('revenue') - col('budget')) # this is still not xecuted (lazy evaluation)
df_with_revenue.show() # xecution happens here 

# Renaming column 
df_movies = df_movies.withColumnRenamed('revenue','total revenue')
df_movies.printSchema()
```
